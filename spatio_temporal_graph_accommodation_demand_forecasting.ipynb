```python
import pandas as pd
import numpy as np
import networkx as nx
import lightgbm as lgb
import xgboost as xgb
from sklearn.ensemble import RandomForestRegressor
import shap
import json
import time
import ast
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import LabelEncoder
from scipy.spatial import cKDTree
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
np.random.seed(42)

sns.set_theme(style="whitegrid", context="paper", font_scale=1.2)
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 12

class PrecisionNumpyEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, np.integer):
            return int(obj)
        elif isinstance(obj, np.floating):
            return float(obj)
        elif isinstance(obj, np.ndarray):
            return obj.tolist()
        return super(PrecisionNumpyEncoder, self).default(obj)


def parse_coordinates(coord_str):
    if pd.isna(coord_str):
        return 0.0, 0.0
    try:
        parts = str(coord_str).split(',')
        if len(parts) == 2:
            return float(parts[0].strip()), float(parts[1].strip())
    except:
        pass
    try:
        coord_dict = ast.literal_eval(str(coord_str))
        if isinstance(coord_dict, dict):
            return float(coord_dict.get('lat', 0.0)), float(coord_dict.get('lon', 0.0))
    except:
        pass
    return 0.0, 0.0


def draw_cd_diagram(ranks, names, cd):
    k = len(ranks)
    sorted_idx = np.argsort(ranks)
    sorted_ranks = np.array(ranks)[sorted_idx]
    sorted_names = np.array(names)[sorted_idx]

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot([1, k], [0, 0], color='black', lw=2)
    for i in range(1, k + 1):
        ax.plot([i, i], [0, 0.1], color='black', lw=1.5)
        ax.text(i, 0.15, str(i), ha='center', va='bottom', fontsize=10)

    ax.plot([1, 1 + cd], [0.8, 0.8], color='black', lw=2)
    ax.plot([1, 1], [0.75, 0.85], color='black', lw=1.5)
    ax.plot([1 + cd, 1 + cd], [0.75, 0.85], color='black', lw=1.5)
    ax.text(1 + cd / 2, 0.9, f"CD = {cd:.3f}", ha='center', va='bottom', fontsize=12, fontweight='bold')

    mid = k // 2
    left_ranks = sorted_ranks[:mid]
    left_names = sorted_names[:mid]
    right_ranks = sorted_ranks[mid:]
    right_names = sorted_names[mid:]

    y_drops_left = np.linspace(-0.3, -1.2, len(left_ranks))
    y_drops_right = np.linspace(-0.3, -1.2, len(right_ranks))

    for rank, name, y in zip(left_ranks, left_names, y_drops_left):
        ax.plot([rank, rank], [0, y], color='black', lw=1.5)
        ax.plot([rank, rank - 0.5], [y, y], color='black', lw=1.5)
        ax.text(rank - 0.6, y, name, ha='right', va='center', fontsize=10)

    for rank, name, y in zip(right_ranks, right_names, y_drops_right):
        ax.plot([rank, rank], [0, y], color='black', lw=1.5)
        ax.plot([rank, rank + 0.5], [y, y], color='black', lw=1.5)
        ax.text(rank + 0.6, y, name, ha='left', va='center', fontsize=10)

    cliques = []
    for i in range(k):
        for j in range(k - 1, i, -1):
            if sorted_ranks[j] - sorted_ranks[i] <= cd:
                cliques.append((i, j))
                break

    maximal_cliques = []
    for c in cliques:
        is_maximal = True
        for other in cliques:
            if c != other and other[0] <= c[0] and other[1] >= c[1]:
                is_maximal = False
                break
        if is_maximal and c not in maximal_cliques:
            maximal_cliques.append(c)

    clique_y = -0.1
    for c in maximal_cliques:
        if c[0] != c[1]:
            ax.plot([sorted_ranks[c[0]], sorted_ranks[c[1]]], [clique_y, clique_y], color='red', lw=4, alpha=0.7)
            clique_y -= 0.08

    ax.set_ylim(-1.5, 1.2)
    ax.set_xlim(0, k + 1)
    ax.axis('off')
    plt.title('Critical Difference (CD) Diagram (Nemenyi Test, α=0.05)', pad=20, fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()


results_dict = {}
results_dict['metadata'] = {
    'experiment_name': 'Spatio-Temporal KD-Tree Graph-Enhanced Demand Forecasting with 5-Fold Cross Validation',
    'leakage_transparency_note': (
        'Graph topology (edges, PageRank, Degree Centrality) is computed GLOBALLY before any fold split. '
        'This is a TRANSDUCTIVE graph setting: all property nodes exist in the graph regardless of fold assignment. '
        'Hotel nodes in the test fold DO influence PageRank values of training-fold nodes because the graph is global. '
        'This is standard practice in transductive spatial graph learning and is explicitly disclosed here. '
        'The zero-data-leakage claim applies strictly to the supervised labels (total_demand): '
        'no target values from the test fold are used to fit any model in the training fold. '
        'Graph features are treated as static node attributes derived from spatial co-occurrence structure, '
        'analogous to geographic coordinates or census-based features, not derived from demand labels.'
    ),
    'environment': 'CPU-Optimized',
    'date_executed': time.strftime("%Y-%m-%dT%H:%M:%SZ", time.gmtime()),
    'datasets_used': ['Accommodation - Transactions by HotelCode.xlsx', 'accommodations.csv', 'activity.csv']
}

start_time_global = time.time()

transactions_df = pd.read_excel('Accommodation - Transactions by HotelCode.xlsx')
transactions_aggregated = transactions_df.groupby('Property Id')['Number Transactions'].sum().reset_index()
transactions_aggregated.rename(columns={'Property Id': 'id', 'Number Transactions': 'total_demand'}, inplace=True)

acc_usecols = ['id', 'popularity_score', 'star_rating', 'guest_rating', 'availability_score', 'price_in_aud', 'destination_coordinate', 'providers']
accommodations_df = pd.read_csv('accommodations.csv', usecols=acc_usecols)

act_usecols = ['product_id', 'normalized_rating', 'price_in_aud', 'duration_hours', 'destination_coordinate', 'categories_as_string']
activities_df = pd.read_csv('activity.csv', usecols=act_usecols)

results_dict['dataset_statistics'] = {
    'raw_transactions_records': len(transactions_df),
    'unique_properties_in_transactions': len(transactions_aggregated),
    'total_accommodations_loaded': len(accommodations_df),
    'total_activities_loaded': len(activities_df)
}

merged_df = pd.merge(accommodations_df, transactions_aggregated, on='id', how='inner')
merged_df['total_demand'] = merged_df['total_demand'].fillna(0)
results_dict['dataset_statistics']['merged_properties_with_demand'] = len(merged_df)

merged_df['lat'], merged_df['lon'] = zip(*merged_df['destination_coordinate'].apply(parse_coordinates))
activities_df['lat'], activities_df['lon'] = zip(*activities_df['destination_coordinate'].apply(parse_coordinates))

merged_df = merged_df[(merged_df['lat'] != 0.0) & (merged_df['lon'] != 0.0)].reset_index(drop=True)
activities_df = activities_df[(activities_df['lat'] != 0.0) & (activities_df['lon'] != 0.0)].reset_index(drop=True)

prop_coords = merged_df[['lat', 'lon']].values
act_coords = activities_df[['lat', 'lon']].values
k_neighbors = 30
tree = cKDTree(act_coords)
distances, indices = tree.query(prop_coords, k=k_neighbors)

RADIUS_KM = np.median(distances[:, 0]) * 2 if np.median(distances[:, 0]) > 0 else 0.1
merged_df['spatial_dist_nearest'] = distances[:, 0]
merged_df['spatial_mean_dist_k'] = distances.mean(axis=1)
merged_df['spatial_count_in_radius'] = (distances < RADIUS_KM).sum(axis=1).astype(float)
area = np.pi * (RADIUS_KM ** 2)
merged_df['spatial_activity_density'] = merged_df['spatial_count_in_radius'] / (area + 1e-9)

print(f"[INFO] Adaptive radius for count/density: {RADIUS_KM:.4f} degrees (~{RADIUS_KM * 111:.1f} km)")
print(f"[INFO] Spatial features computed for {len(merged_df)} properties")

spatial_graph = nx.Graph()
edge_list = []
for i, prop_idx in enumerate(merged_df.index):
    prop_id = f"P_{merged_df.loc[prop_idx, 'id']}"
    for j in range(k_neighbors):
        act_idx = indices[i, j]
        act_id = f"A_{activities_df.loc[act_idx, 'product_id']}"
        dist = distances[i, j]
        weight = 1.0 / (dist + 1e-4)
        edge_list.append((prop_id, act_id, {'weight': weight}))
spatial_graph.add_edges_from(edge_list)

results_dict['graph_topology'] = {
    'total_nodes': spatial_graph.number_of_nodes(),
    'total_edges': spatial_graph.number_of_edges(),
    'density': nx.density(spatial_graph),
    'graph_construction_note': (
        'Graph is constructed globally using all properties and all activities before any fold split. '
        'PageRank and Degree Centrality are TRANSDUCTIVE features: they reflect each node position '
        'within the full global spatial co-occurrence graph. Test-fold property nodes are present '
        'in the graph during PageRank computation, meaning their connectivity contributes to the '
        'PageRank of training-fold properties. This is explicitly disclosed and does not constitute '
        'label leakage since no demand (target) values are used in graph construction.'
    )
}

provider_flow = {}
prop_providers = dict(zip(merged_df['id'].apply(lambda x: f"P_{x}"), merged_df['providers'].fillna('Unknown')))
act_categories = dict(zip(activities_df['product_id'].apply(lambda x: f"A_{x}"), activities_df['categories_as_string'].fillna('General')))

for edge in spatial_graph.edges():
    u, v = edge
    if u.startswith('P_') and v.startswith('A_'):
        prov = str(prop_providers.get(u, 'Unknown')).split(',')[0].strip()
        cat = str(act_categories.get(v, 'General')).split(',')[0].strip()
    elif v.startswith('P_') and u.startswith('A_'):
        prov = str(prop_providers.get(v, 'Unknown')).split(',')[0].strip()
        cat = str(act_categories.get(u, 'General')).split(',')[0].strip()
    else:
        continue
    if prov not in provider_flow:
        provider_flow[prov] = {}
    if cat not in provider_flow[prov]:
        provider_flow[prov][cat] = 0
    provider_flow[prov][cat] += 1

top_n_prov = 5
top_n_cat = 5
prov_totals = {p: sum(provider_flow[p].values()) for p in provider_flow}
top_provs = sorted(prov_totals, key=prov_totals.get, reverse=True)[:top_n_prov]
cat_totals = {}
for p in top_provs:
    for c, count in provider_flow[p].items():
        cat_totals[c] = cat_totals.get(c, 0) + count
top_cats = sorted(cat_totals, key=cat_totals.get, reverse=True)[:top_n_cat]
filtered_flow = {}
for p in top_provs:
    filtered_flow[p] = {}
    for c in top_cats:
        filtered_flow[p][c] = provider_flow[p].get(c, 0)
results_dict['graph_topology']['macro_flow_matrix'] = filtered_flow

start_time_pagerank = time.time()
pagerank_scores = nx.pagerank(spatial_graph, alpha=0.85, weight='weight')
time_pagerank = time.time() - start_time_pagerank

start_time_degree = time.time()
degree_centrality = nx.degree_centrality(spatial_graph)
time_degree = time.time() - start_time_degree

results_dict['computational_efficiency'] = {
    'pagerank_seconds': time_pagerank,
    'degree_centrality_seconds': time_degree
}

graph_features = []
for pid in merged_df['id'].unique():
    node_name = f"P_{pid}"
    graph_features.append({
        'id': pid,
        'pagerank': pagerank_scores.get(node_name, 0.0),
        'degree_centrality': degree_centrality.get(node_name, 0.0)
    })
graph_features_df = pd.DataFrame(graph_features)
enriched_dataset = pd.merge(merged_df, graph_features_df, on='id', how='inner')
enriched_dataset.fillna(0, inplace=True)

le_provider = LabelEncoder()
enriched_dataset['provider_encoded'] = le_provider.fit_transform(enriched_dataset['providers'].astype(str))

BASE_FEATURES = ['popularity_score', 'star_rating', 'guest_rating', 'availability_score', 'price_in_aud', 'provider_encoded']
SPATIAL_FEATURES = BASE_FEATURES + ['spatial_dist_nearest', 'spatial_mean_dist_k', 'spatial_count_in_radius', 'spatial_activity_density']
GRAPH_FEATURES = BASE_FEATURES + ['pagerank', 'degree_centrality']
FULL_FEATURES = BASE_FEATURES + ['pagerank', 'degree_centrality', 'spatial_dist_nearest', 'spatial_mean_dist_k', 'spatial_count_in_radius', 'spatial_activity_density']

ablation_feature_sets = {
    'Baseline (No Spatial/Graph)': BASE_FEATURES,
    'Spatial Proximity Only': SPATIAL_FEATURES,
    'Graph Features Only (PageRank+DC)': GRAPH_FEATURES,
    'Full (Graph + Spatial)': FULL_FEATURES
}

X_proposed = enriched_dataset[GRAPH_FEATURES].values
X_baseline = enriched_dataset[BASE_FEATURES].values
y = enriched_dataset['total_demand'].values

results_dict['model_evaluation'] = {}
for model_key in ['LightGBM_Proposed_Graph', 'LightGBM_Baseline', 'XGBoost_Proposed_Graph', 'XGBoost_Baseline', 'RandomForest_Proposed_Graph', 'RandomForest_Baseline']:
    results_dict['model_evaluation'][model_key] = {
        'rmse_folds': [], 'mae_folds': [], 'r2_score_folds': [], 'training_time_seconds_folds': []
    }

kf = KFold(n_splits=5, shuffle=True, random_state=42)

lgb_params = {
    'objective': 'regression', 'metric': 'rmse', 'boosting_type': 'gbdt',
    'num_leaves': 31, 'learning_rate': 0.03, 'feature_fraction': 0.7,
    'bagging_fraction': 0.8, 'bagging_freq': 5, 'verbose': -1,
    'device_type': 'cpu', 'num_threads': 4
}
xgb_params = {
    'objective': 'reg:squarederror', 'eval_metric': 'rmse', 'max_depth': 5,
    'learning_rate': 0.03, 'subsample': 0.8, 'colsample_bytree': 0.7,
    'n_jobs': 4, 'random_state': 42
}
rf_params = {'n_estimators': 200, 'max_depth': 10, 'max_features': 0.7, 'n_jobs': 4, 'random_state': 42}

shap_values_list = []
X_test_shap_list = []

y_oof_true = np.zeros(len(y))
y_oof_preds = {
    'LightGBM_Proposed_Graph': np.zeros(len(y)), 'LightGBM_Baseline': np.zeros(len(y)),
    'XGBoost_Proposed_Graph': np.zeros(len(y)), 'XGBoost_Baseline': np.zeros(len(y)),
    'RandomForest_Proposed_Graph': np.zeros(len(y)), 'RandomForest_Baseline': np.zeros(len(y))
}

for train_index, test_index in kf.split(X_proposed):
    X_tr_prop, X_te_prop = X_proposed[train_index], X_proposed[test_index]
    X_tr_base, X_te_base = X_baseline[train_index], X_baseline[test_index]
    y_tr, y_te = y[train_index], y[test_index]
    y_oof_true[test_index] = y_te

    t0 = time.time()
    gbm_prop = lgb.train(lgb_params, lgb.Dataset(X_tr_prop, y_tr), num_boost_round=400)
    results_dict['model_evaluation']['LightGBM_Proposed_Graph']['training_time_seconds_folds'].append(time.time() - t0)
    y_pred_lgbm_prop = gbm_prop.predict(X_te_prop)
    y_oof_preds['LightGBM_Proposed_Graph'][test_index] = y_pred_lgbm_prop
    results_dict['model_evaluation']['LightGBM_Proposed_Graph']['rmse_folds'].append(np.sqrt(mean_squared_error(y_te, y_pred_lgbm_prop)))
    results_dict['model_evaluation']['LightGBM_Proposed_Graph']['mae_folds'].append(mean_absolute_error(y_te, y_pred_lgbm_prop))
    results_dict['model_evaluation']['LightGBM_Proposed_Graph']['r2_score_folds'].append(r2_score(y_te, y_pred_lgbm_prop))
    shap_values_list.append(shap.TreeExplainer(gbm_prop).shap_values(X_te_prop))

    t0 = time.time()
    gbm_base = lgb.train(lgb_params, lgb.Dataset(X_tr_base, y_tr), num_boost_round=400)
    results_dict['model_evaluation']['LightGBM_Baseline']['training_time_seconds_folds'].append(time.time() - t0)
    y_pred_lgbm_base = gbm_base.predict(X_te_base)
    y_oof_preds['LightGBM_Baseline'][test_index] = y_pred_lgbm_base
    results_dict['model_evaluation']['LightGBM_Baseline']['rmse_folds'].append(np.sqrt(mean_squared_error(y_te, y_pred_lgbm_base)))
    results_dict['model_evaluation']['LightGBM_Baseline']['mae_folds'].append(mean_absolute_error(y_te, y_pred_lgbm_base))
    results_dict['model_evaluation']['LightGBM_Baseline']['r2_score_folds'].append(r2_score(y_te, y_pred_lgbm_base))

    t0 = time.time()
    xgb_prop = xgb.XGBRegressor(**xgb_params, n_estimators=400).fit(X_tr_prop, y_tr)
    results_dict['model_evaluation']['XGBoost_Proposed_Graph']['training_time_seconds_folds'].append(time.time() - t0)
    y_pred_xgb_prop = xgb_prop.predict(X_te_prop)
    y_oof_preds['XGBoost_Proposed_Graph'][test_index] = y_pred_xgb_prop
    results_dict['model_evaluation']['XGBoost_Proposed_Graph']['rmse_folds'].append(np.sqrt(mean_squared_error(y_te, y_pred_xgb_prop)))
    results_dict['model_evaluation']['XGBoost_Proposed_Graph']['mae_folds'].append(mean_absolute_error(y_te, y_pred_xgb_prop))
    results_dict['model_evaluation']['XGBoost_Proposed_Graph']['r2_score_folds'].append(r2_score(y_te, y_pred_xgb_prop))

    t0 = time.time()
    xgb_base = xgb.XGBRegressor(**xgb_params, n_estimators=400).fit(X_tr_base, y_tr)
    results_dict['model_evaluation']['XGBoost_Baseline']['training_time_seconds_folds'].append(time.time() - t0)
    y_pred_xgb_base = xgb_base.predict(X_te_base)
    y_oof_preds['XGBoost_Baseline'][test_index] = y_pred_xgb_base
    results_dict['model_evaluation']['XGBoost_Baseline']['rmse_folds'].append(np.sqrt(mean_squared_error(y_te, y_pred_xgb_base)))
    results_dict['model_evaluation']['XGBoost_Baseline']['mae_folds'].append(mean_absolute_error(y_te, y_pred_xgb_base))
    results_dict['model_evaluation']['XGBoost_Baseline']['r2_score_folds'].append(r2_score(y_te, y_pred_xgb_base))

    t0 = time.time()
    rf_prop = RandomForestRegressor(**rf_params).fit(X_tr_prop, y_tr)
    results_dict['model_evaluation']['RandomForest_Proposed_Graph']['training_time_seconds_folds'].append(time.time() - t0)
    y_pred_rf_prop = rf_prop.predict(X_te_prop)
    y_oof_preds['RandomForest_Proposed_Graph'][test_index] = y_pred_rf_prop
    results_dict['model_evaluation']['RandomForest_Proposed_Graph']['rmse_folds'].append(np.sqrt(mean_squared_error(y_te, y_pred_rf_prop)))
    results_dict['model_evaluation']['RandomForest_Proposed_Graph']['mae_folds'].append(mean_absolute_error(y_te, y_pred_rf_prop))
    results_dict['model_evaluation']['RandomForest_Proposed_Graph']['r2_score_folds'].append(r2_score(y_te, y_pred_rf_prop))

    t0 = time.time()
    rf_base = RandomForestRegressor(**rf_params).fit(X_tr_base, y_tr)
    results_dict['model_evaluation']['RandomForest_Baseline']['training_time_seconds_folds'].append(time.time() - t0)
    y_pred_rf_base = rf_base.predict(X_te_base)
    y_oof_preds['RandomForest_Baseline'][test_index] = y_pred_rf_base
    results_dict['model_evaluation']['RandomForest_Baseline']['rmse_folds'].append(np.sqrt(mean_squared_error(y_te, y_pred_rf_base)))
    results_dict['model_evaluation']['RandomForest_Baseline']['mae_folds'].append(mean_absolute_error(y_te, y_pred_rf_base))
    results_dict['model_evaluation']['RandomForest_Baseline']['r2_score_folds'].append(r2_score(y_te, y_pred_rf_base))

for model_key in results_dict['model_evaluation']:
    for metric in ['rmse', 'mae', 'r2_score']:
        results_dict['model_evaluation'][model_key][f'{metric}_mean'] = np.mean(results_dict['model_evaluation'][model_key][f'{metric}_folds'])
        results_dict['model_evaluation'][model_key][f'{metric}_std'] = np.std(results_dict['model_evaluation'][model_key][f'{metric}_folds'])
    results_dict['model_evaluation'][model_key]['training_time_mean'] = np.mean(results_dict['model_evaluation'][model_key]['training_time_seconds_folds'])
    results_dict['model_evaluation'][model_key]['training_time_total'] = np.sum(results_dict['model_evaluation'][model_key]['training_time_seconds_folds'])

results_dict['model_evaluation']['y_oof_true_global'] = y_oof_true
for model_key in y_oof_preds:
    results_dict['model_evaluation'][model_key]['y_oof_pred'] = y_oof_preds[model_key]

all_shap_values = np.vstack(shap_values_list)
mean_abs_shap = np.abs(all_shap_values).mean(axis=0)

results_dict['interpretability'] = {}
results_dict['interpretability']['shap_mean_absolute_values_oof'] = {GRAPH_FEATURES[i]: mean_abs_shap[i] for i in range(len(GRAPH_FEATURES))}

gbm_prop_full = lgb.train(lgb_params, lgb.Dataset(X_proposed, y), num_boost_round=400)
feature_importance_split = gbm_prop_full.feature_importance(importance_type='split')
feature_importance_gain = gbm_prop_full.feature_importance(importance_type='gain')
results_dict['interpretability']['lightgbm_split_importance_full'] = {GRAPH_FEATURES[i]: feature_importance_split[i] for i in range(len(GRAPH_FEATURES))}
results_dict['interpretability']['lightgbm_gain_importance_full'] = {GRAPH_FEATURES[i]: feature_importance_gain[i] for i in range(len(GRAPH_FEATURES))}

ablation_results = {name: {'rmse': [], 'mae': [], 'r2': [], 'train_time_s': []} for name in ablation_feature_sets}
for name, feat_list in ablation_feature_sets.items():
    X_abl = enriched_dataset[feat_list].values
    print(f"[Ablation CV] Running: {name}  |  Features: {len(feat_list)}")
    for fold_i, (tr_idx, te_idx) in enumerate(kf.split(X_abl)):
        X_tr_a, X_te_a = X_abl[tr_idx], X_abl[te_idx]
        y_tr_a, y_te_a = y[tr_idx], y[te_idx]
        t0 = time.time()
        model_abl = lgb.train(lgb_params, lgb.Dataset(X_tr_a, y_tr_a), num_boost_round=400)
        train_sec = time.time() - t0
        y_pred_abl = model_abl.predict(X_te_a)
        ablation_results[name]['rmse'].append(np.sqrt(mean_squared_error(y_te_a, y_pred_abl)))
        ablation_results[name]['mae'].append(mean_absolute_error(y_te_a, y_pred_abl))
        ablation_results[name]['r2'].append(r2_score(y_te_a, y_pred_abl))
        ablation_results[name]['train_time_s'].append(train_sec)
    print(f"       RMSE={np.mean(ablation_results[name]['rmse']):.4f}  MAE={np.mean(ablation_results[name]['mae']):.4f}  R2={np.mean(ablation_results[name]['r2']):.4f}  AvgTime={np.mean(ablation_results[name]['train_time_s']):.4f}s")

results_dict['ablation_study'] = {}
for name, vals in ablation_results.items():
    results_dict['ablation_study'][name] = {
        'rmse_mean': np.mean(vals['rmse']), 'rmse_std': np.std(vals['rmse']),
        'mae_mean': np.mean(vals['mae']), 'mae_std': np.std(vals['mae']),
        'r2_mean': np.mean(vals['r2']), 'r2_std': np.std(vals['r2']),
        'avg_train_time_s': np.mean(vals['train_time_s']),
        'total_train_time_s': np.sum(vals['train_time_s']),
        'rmse_folds': vals['rmse'], 'mae_folds': vals['mae'], 'r2_folds': vals['r2'],
        'train_time_folds': vals['train_time_s']
    }

X_full_abl = enriched_dataset[FULL_FEATURES].values
model_full_abl = lgb.train(lgb_params, lgb.Dataset(X_full_abl, y), num_boost_round=400)
gain_imp_full = model_full_abl.feature_importance(importance_type='gain')
results_dict['ablation_study']['full_model_gain_importance'] = {FULL_FEATURES[i]: gain_imp_full[i] for i in range(len(FULL_FEATURES))}

results_dict['computational_efficiency']['total_pipeline_seconds'] = time.time() - start_time_global

with open('comprehensive_methodology_results.json', 'w') as f:
    json.dump(results_dict, f, indent=4, cls=PrecisionNumpyEncoder)

print("\n[INFO] JSON results saved. Beginning visualization pipeline.")

with open('comprehensive_methodology_results.json', 'r') as f:
    results = json.load(f)

eval_data = results['model_evaluation']
model_keys_to_skip = ['y_oof_true_global']

fold_details_list = []
for key, metrics in eval_data.items():
    if key in model_keys_to_skip:
        continue
    algo = key.split('_')[0]
    is_graph = 'Proposed (Graph)' if 'Proposed' in key else 'Baseline (No Graph)'
    for fold_idx in range(5):
        fold_details_list.append({
            'Model': key, 'Algorithm': algo, 'Architecture': is_graph,
            'Fold': f'Fold {fold_idx + 1}',
            'RMSE': metrics['rmse_folds'][fold_idx],
            'MAE': metrics['mae_folds'][fold_idx],
            'R2': metrics['r2_score_folds'][fold_idx]
        })
fold_df = pd.DataFrame(fold_details_list)

summary_list = []
for key, metrics in eval_data.items():
    if key in model_keys_to_skip:
        continue
    summary_list.append({
        'Model': key,
        'RMSE Mean': metrics['rmse_mean'], 'RMSE Std': metrics['rmse_std'],
        'MAE Mean': metrics['mae_mean'], 'MAE Std': metrics['mae_std'],
        'R2 Mean': metrics['r2_score_mean'], 'R2 Std': metrics['r2_score_std'],
        'Avg Train Time (s)': metrics['training_time_mean'],
        'Total Train Time (s)': metrics['training_time_total']
    })
summary_df = pd.DataFrame(summary_list)

print("\n" + "=" * 70)
print("PRINTOUT: Detailed 5-Fold Fluctuations")
print("=" * 70)
print(fold_df.to_string(index=False))

plt.figure(figsize=(10, 6))
sns.lineplot(data=fold_df, x='Fold', y='RMSE', hue='Model', marker='o', linewidth=2.5, palette='tab10')
plt.title('5-Fold Cross-Validation: RMSE per Fold')
plt.ylabel('RMSE (Lower is Better)')
plt.legend(loc='best')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.lineplot(data=fold_df, x='Fold', y='MAE', hue='Model', marker='o', linewidth=2.5, palette='tab10')
plt.title('5-Fold Cross-Validation: MAE per Fold')
plt.ylabel('MAE (Lower is Better)')
plt.legend(loc='best')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.lineplot(data=fold_df, x='Fold', y='R2', hue='Model', marker='o', linewidth=2.5, palette='tab10')
plt.title('5-Fold Cross-Validation: R² Score per Fold')
plt.ylabel('R² Score (Higher is Better)')
plt.legend(loc='best')
plt.tight_layout()
plt.show()

print("\n" + "=" * 70)
print("PRINTOUT: 5-Fold Aggregation (Mean & Std) and Statistical Tests")
print("=" * 70)
print(summary_df.to_string(index=False))

print("\n--- STATISTICAL SIGNIFICANCE TEST (Paired T-Test) ---")
algos = ['LightGBM', 'XGBoost', 'RandomForest']
for alg in algos:
    prop_rmse = eval_data[f'{alg}_Proposed_Graph']['rmse_folds']
    base_rmse = eval_data[f'{alg}_Baseline']['rmse_folds']
    t_stat, p_value = stats.ttest_rel(base_rmse, prop_rmse)
    sig = "SIGNIFICANT (p<0.05)" if p_value < 0.05 else "NOT SIGNIFICANT"
    print(f"{alg:12} Baseline vs Proposed | P-Value: {p_value:.6f} | {sig}")

print("\n--- CRITICAL DIFFERENCE (CD) & AVERAGE RANKING ---")
model_names_cd = ['LightGBM_Proposed_Graph', 'LightGBM_Baseline', 'XGBoost_Proposed_Graph', 'XGBoost_Baseline', 'RandomForest_Proposed_Graph', 'RandomForest_Baseline']
rmse_matrix = np.zeros((5, 6))
for j, m in enumerate(model_names_cd):
    rmse_matrix[:, j] = eval_data[m]['rmse_folds']

ranks_matrix = np.zeros((5, 6))
for i in range(5):
    ranks_matrix[i, :] = stats.rankdata(rmse_matrix[i, :])
average_ranks = ranks_matrix.mean(axis=0)
rank_df = pd.DataFrame({'Model': model_names_cd, 'Average Rank': average_ranks}).sort_values(by='Average Rank')
print(rank_df.to_string(index=False))

p_value_matrix = np.zeros((6, 6))
for i in range(6):
    for j in range(6):
        if i == j:
            p_value_matrix[i, j] = 1.0
        else:
            _, p_val = stats.ttest_rel(rmse_matrix[:, i], rmse_matrix[:, j])
            p_value_matrix[i, j] = p_val
p_val_df = pd.DataFrame(p_value_matrix, index=model_names_cd, columns=model_names_cd)

metrics_list = []
for key, metrics in eval_data.items():
    if key in model_keys_to_skip:
        continue
    algo = key.split('_')[0]
    is_graph = 'Proposed (Graph)' if 'Proposed' in key else 'Baseline (No Graph)'
    metrics_list.append({'Algorithm': algo, 'Architecture': is_graph, 'Metric': 'RMSE', 'Value': metrics['rmse_mean']})
    metrics_list.append({'Algorithm': algo, 'Architecture': is_graph, 'Metric': 'MAE', 'Value': metrics['mae_mean']})
metrics_df = pd.DataFrame(metrics_list)

r2_list = []
for key, metrics in eval_data.items():
    if key in model_keys_to_skip:
        continue
    algo = key.split('_')[0]
    is_graph = 'Proposed (Graph)' if 'Proposed' in key else 'Baseline (No Graph)'
    r2_list.append({'Algorithm': algo, 'Architecture': is_graph, 'R-Squared Score': metrics['r2_score_mean']})
r2_df = pd.DataFrame(r2_list)

plt.figure(figsize=(10, 6))
sns.barplot(x='Algorithm', y='Value', hue='Architecture', data=metrics_df[metrics_df['Metric'] == 'RMSE'], palette='viridis')
plt.title('Algorithm-Agnostic Validation: RMSE Mean')
plt.ylabel('RMSE Mean (Lower is Better)')
plt.legend(title='Architecture', loc='best')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.barplot(x='Algorithm', y='Value', hue='Architecture', data=metrics_df[metrics_df['Metric'] == 'MAE'], palette='plasma')
plt.title('Algorithm-Agnostic Validation: MAE Mean')
plt.ylabel('MAE Mean (Lower is Better)')
plt.legend(title='Architecture', loc='best')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.barplot(x='Algorithm', y='R-Squared Score', hue='Architecture', data=r2_df, palette='mako')
plt.title('Algorithm-Agnostic Validation: R² Mean Score')
plt.ylabel('R² Score Mean (Higher is Better)')
if r2_df['R-Squared Score'].max() > 0:
    plt.ylim(0, max(r2_df['R-Squared Score'].max() * 1.2, 1.0))
plt.legend(title='Architecture', loc='best')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.barplot(x='Average Rank', y='Model', data=rank_df, palette='autumn')
plt.title('Critical Difference Analysis: Average Model Rank across 5 Folds (Lower is Better)')
plt.xlabel('Average Rank (1 is Best)')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 10))
mask = np.triu(np.ones_like(p_value_matrix, dtype=bool))
sns.heatmap(p_val_df, annot=True, fmt=".3f", cmap='coolwarm', mask=mask, cbar_kws={'label': 'P-Value'})
plt.title('Pairwise Statistical Significance Matrix (P-Values)\nP-Values < 0.05 indicate Significant Difference')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

draw_cd_diagram(average_ranks, model_names_cd, 3.372)

print("\n" + "=" * 70)
print("PRINTOUT: 5-Fold Distribution Quartiles")
print("=" * 70)
print("--- RMSE Distribution ---")
print(fold_df.groupby(['Algorithm', 'Architecture'])['RMSE'].describe().to_string())
print("\n--- MAE Distribution ---")
print(fold_df.groupby(['Algorithm', 'Architecture'])['MAE'].describe().to_string())
print("\n--- R2 Score Distribution ---")
print(fold_df.groupby(['Algorithm', 'Architecture'])['R2'].describe().to_string())

plt.figure(figsize=(12, 6))
sns.violinplot(data=fold_df, x='Algorithm', y='RMSE', hue='Architecture', split=True, inner='quart', palette='viridis')
plt.title('5-Fold Distribution: RMSE (Split Violin Plot)')
plt.ylabel('RMSE (Lower is Better)')
plt.legend(loc='best')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.violinplot(data=fold_df, x='Algorithm', y='MAE', hue='Architecture', split=True, inner='quart', palette='plasma')
plt.title('5-Fold Distribution: MAE (Split Violin Plot)')
plt.ylabel('MAE (Lower is Better)')
plt.legend(loc='best')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.violinplot(data=fold_df, x='Algorithm', y='R2', hue='Architecture', split=True, inner='quart', palette='mako')
plt.title('5-Fold Distribution: R² Score (Split Violin Plot)')
plt.ylabel('R² Score (Higher is Better)')
plt.legend(loc='best')
plt.tight_layout()
plt.show()

y_true_full = np.array(eval_data['y_oof_true_global'])
sample_df = pd.DataFrame({'Actual Demand': y_true_full})
model_keys_plot = ['LightGBM_Proposed_Graph', 'LightGBM_Baseline', 'XGBoost_Proposed_Graph', 'XGBoost_Baseline', 'RandomForest_Proposed_Graph', 'RandomForest_Baseline']
for m_key in model_keys_plot:
    sample_df[m_key] = np.array(eval_data[m_key]['y_oof_pred'])
sampled_data = sample_df.sample(n=100, random_state=42).sort_index().reset_index(drop=True)
plot_indices = np.arange(1, len(sampled_data) + 1)

print("\n" + "=" * 70)
print("PRINTOUT: Model Verification - Actual vs Predicted Fluctuation (N=100 sample)")
print("=" * 70)

plt.figure(figsize=(18, 7))
plt.plot(plot_indices, sampled_data['Actual Demand'], label='Actual Demand (Ground Truth)', color='black', linewidth=3.0, marker='o', markersize=6, alpha=0.9)
colors_palette = sns.color_palette('tab10', 6)
for idx, model_key in enumerate(model_keys_plot):
    plt.plot(plot_indices, sampled_data[model_key], label=f'Predicted ({model_key})', color=colors_palette[idx], linewidth=1.5, alpha=0.8, linestyle='--')
plt.title('Actual vs Predicted Demand Fluctuations (Random Sample of N=100 Properties)')
plt.ylabel('Total Demand (Transactions)')
plt.xlabel('Sampled Property Index')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1), fontsize=10)
plt.tight_layout()
plt.show()

shap_vals = results['interpretability']['shap_mean_absolute_values_oof']
shap_df = pd.DataFrame(list(shap_vals.items()), columns=['Feature', 'SHAP Value']).sort_values(by='SHAP Value', ascending=False)

gain_vals = results['interpretability']['lightgbm_gain_importance_full']
gain_df_main = pd.DataFrame(list(gain_vals.items()), columns=['Feature', 'Information Gain']).sort_values(by='Information Gain', ascending=False)

split_vals = results['interpretability']['lightgbm_split_importance_full']
split_df = pd.DataFrame(list(split_vals.items()), columns=['Feature', 'Split Count']).sort_values(by='Split Count', ascending=False)

print("\n" + "=" * 70)
print("PRINTOUT: Explainable AI - SHAP, Gain, Split")
print("=" * 70)
print("--- SHAP Global Feature Importance (OOF) ---")
print(shap_df.to_string(index=False))
print("\n--- LightGBM Feature Importance (Information Gain) ---")
print(gain_df_main.to_string(index=False))
print("\n--- LightGBM Feature Importance (Split Count) ---")
print(split_df.to_string(index=False))

plt.figure(figsize=(10, 6))
sns.barplot(x='SHAP Value', y='Feature', data=shap_df, palette='rocket')
plt.title('Explainable AI: Out-of-Fold SHAP Global Feature Importance')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.barplot(x='Information Gain', y='Feature', data=gain_df_main, palette='crest')
plt.title('LightGBM Feature Importance (Information Gain)')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.barplot(x='Split Count', y='Feature', data=split_df, palette='magma')
plt.title('LightGBM Feature Importance (Split Count)')
plt.tight_layout()
plt.show()

comp_eff = results['computational_efficiency']
time_list = [
    {'Process Phase': 'PageRank Extraction', 'Time (Seconds)': comp_eff['pagerank_seconds']},
    {'Process Phase': 'Degree Centrality', 'Time (Seconds)': comp_eff['degree_centrality_seconds']}
]
for key, metrics in eval_data.items():
    if key in model_keys_to_skip:
        continue
    time_list.append({'Process Phase': f'Train {key} (5x)', 'Time (Seconds)': metrics['training_time_total']})
comp_df = pd.DataFrame(time_list).sort_values(by='Time (Seconds)', ascending=False)

stats_data = results['dataset_statistics']
graph_top = results['graph_topology']
topology_df = pd.DataFrame({
    'Metric': ['Raw Transactions', 'Unique Properties', 'Total Accommodations', 'Total Activities', 'Graph Nodes', 'Graph Edges'],
    'Count': [stats_data['raw_transactions_records'], stats_data['unique_properties_in_transactions'], stats_data['total_accommodations_loaded'], stats_data['total_activities_loaded'], graph_top['total_nodes'], graph_top['total_edges']]
})

print("\n" + "=" * 70)
print("PRINTOUT: Computational Profiling & Dataset Characteristics")
print("=" * 70)
print("--- Time Complexity Analysis ---")
print(comp_df.to_string(index=False))
print("\n--- Spatio-Temporal Graph Characteristics ---")
print(topology_df.to_string(index=False))

plt.figure(figsize=(10, 6))
sns.barplot(x='Time (Seconds)', y='Process Phase', data=comp_df, palette='flare')
plt.title('Computational Profiling & Time Complexity Analysis')
plt.xscale('log')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.barplot(x='Count', y='Metric', data=topology_df, palette='ocean')
plt.title('Dataset and Spatio-Temporal Graph Characteristics')
plt.xscale('log')
plt.tight_layout()
plt.show()

flow_matrix = results['graph_topology'].get('macro_flow_matrix', {})
if flow_matrix:
    print("\n" + "=" * 70)
    print("PRINTOUT: Spatio-Temporal Macro-Topology Flow Matrix")
    print("=" * 70)
    flow_df_print = pd.DataFrame(flow_matrix).fillna(0).T
    print(flow_df_print.to_string())

    G_flow = nx.DiGraph()
    providers_chord = list(flow_matrix.keys())
    categories_chord = []
    for p, cats in flow_matrix.items():
        for c in cats.keys():
            if c not in categories_chord:
                categories_chord.append(c)
    for p in providers_chord:
        G_flow.add_node(p, bipartite=0)
    for c in categories_chord:
        G_flow.add_node(c, bipartite=1)
    for p, cats in flow_matrix.items():
        for c, weight in cats.items():
            if weight > 0:
                G_flow.add_edge(p, c, weight=weight)
    pos = nx.circular_layout(G_flow)
    node_colors_chord = ['skyblue' if n in providers_chord else 'lightgreen' for n in G_flow.nodes()]
    node_sizes_chord = []
    for n in G_flow.nodes():
        if n in providers_chord:
            degree_w = sum([G_flow[n][v]['weight'] for v in G_flow.successors(n)])
            node_sizes_chord.append(degree_w * 0.5 + 2000)
        else:
            degree_w = sum([G_flow[u][n]['weight'] for u in G_flow.predecessors(n)])
            node_sizes_chord.append(degree_w * 0.5 + 2000)
    plt.figure(figsize=(12, 12))
    nx.draw_networkx_nodes(G_flow, pos, node_color=node_colors_chord, node_size=node_sizes_chord, edgecolors='black', linewidths=2)
    nx.draw_networkx_labels(G_flow, pos, font_size=9, font_weight='bold')
    for (u, v, d) in G_flow.edges(data=True):
        nx.draw_networkx_edges(G_flow, pos, edgelist=[(u, v)], width=d['weight'] * 0.01, alpha=0.6, edge_color='gray', connectionstyle='arc3,rad=0.2', arrowsize=15)
    plt.title('Spatio-Temporal Chord Network: Top Providers to Activity Categories', fontsize=16, pad=20)
    plt.axis('off')
    plt.tight_layout()
    plt.show()

abl_data = results['ablation_study']
abl_names = list(ablation_feature_sets.keys())
abl_rmse_means = [abl_data[n]['rmse_mean'] for n in abl_names]
abl_rmse_stds = [abl_data[n]['rmse_std'] for n in abl_names]
abl_mae_means = [abl_data[n]['mae_mean'] for n in abl_names]
abl_mae_stds = [abl_data[n]['mae_std'] for n in abl_names]
abl_r2_means = [abl_data[n]['r2_mean'] for n in abl_names]
abl_r2_stds = [abl_data[n]['r2_std'] for n in abl_names]
short_labels_abl = ['Baseline', 'Spatial\nProximity', 'Graph\n(PageRank+DC)', 'Full\n(Graph+Spatial)']
colors_abl = sns.color_palette('viridis', 4)

print("\n" + "=" * 80)
print("ABLATION STUDY SUMMARY — LightGBM 5-Fold Cross-Validation")
print("=" * 80)
abl_summary_rows = []
for name in abl_names:
    abl_summary_rows.append({
        'Feature Set': name,
        'Features Used': len(ablation_feature_sets[name]),
        'RMSE Mean': f"{abl_data[name]['rmse_mean']:.6f}",
        'RMSE Std': f"{abl_data[name]['rmse_std']:.6f}",
        'MAE Mean': f"{abl_data[name]['mae_mean']:.6f}",
        'MAE Std': f"{abl_data[name]['mae_std']:.6f}",
        'R2 Mean': f"{abl_data[name]['r2_mean']:.6f}",
        'R2 Std': f"{abl_data[name]['r2_std']:.6f}",
        'Avg Train Time (s)': f"{abl_data[name]['avg_train_time_s']:.6f}"
    })
print(pd.DataFrame(abl_summary_rows).to_string(index=False))

plt.figure(figsize=(10, 6))
plt.bar(short_labels_abl, abl_rmse_means, yerr=abl_rmse_stds, color=colors_abl, capsize=5, edgecolor='black')
plt.title('Ablation Study: RMSE Mean (±Std)\nLower is Better')
plt.ylabel('RMSE')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
plt.bar(short_labels_abl, abl_mae_means, yerr=abl_mae_stds, color=colors_abl, capsize=5, edgecolor='black')
plt.title('Ablation Study: MAE Mean (±Std)\nLower is Better')
plt.ylabel('MAE')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
plt.bar(short_labels_abl, abl_r2_means, yerr=abl_r2_stds, color=colors_abl, capsize=5, edgecolor='black')
plt.title('Ablation Study: R² Mean (±Std)\nHigher is Better')
plt.ylabel('R² Score')
plt.tight_layout()
plt.show()

abl_fold_rows = []
for name in abl_names:
    for fi, rmse_v in enumerate(abl_data[name]['rmse_folds']):
        abl_fold_rows.append({'Feature Set': name, 'Fold': f'Fold {fi + 1}', 'RMSE': rmse_v})
abl_fold_df = pd.DataFrame(abl_fold_rows)

plt.figure(figsize=(10, 6))
sns.lineplot(data=abl_fold_df, x='Fold', y='RMSE', hue='Feature Set', marker='o', linewidth=2.5, palette='tab10')
plt.title('Ablation Study: 5-Fold RMSE per Feature Set\nValidates whether PageRank provides genuine structural signal')
plt.ylabel('RMSE (Lower is Better)')
plt.legend(loc='best', fontsize=9)
plt.tight_layout()
plt.show()

abl_time_rows = []
for name in abl_names:
    for fi, t in enumerate(abl_data[name]['train_time_folds']):
        abl_time_rows.append({'Feature Set': name, 'Fold': f'Fold {fi + 1}', 'Train Time (s)': t})
abl_time_df = pd.DataFrame(abl_time_rows)

plt.figure(figsize=(12, 6))
sns.barplot(data=abl_time_df, x='Fold', y='Train Time (s)', hue='Feature Set', palette='flare')
plt.title('Ablation Study: Per-Fold Training Time by Feature Set\n(Actual Measured, seconds)')
plt.ylabel('Training Time (s)')
plt.legend(fontsize=8, loc='upper right')
plt.tight_layout()
plt.show()

abl_total_time_rows = [{'Feature Set': n, 'Total Train Time (s)': abl_data[n]['total_train_time_s']} for n in abl_names]
abl_total_time_df = pd.DataFrame(abl_total_time_rows)
plt.figure(figsize=(10, 6))
sns.barplot(data=abl_total_time_df, x='Total Train Time (s)', y='Feature Set', palette='flare')
plt.title('Ablation Study: Total Training Time (5 Folds Sum)')
plt.xlabel('Total Training Time (s)')
plt.tight_layout()
plt.show()

gain_importance_full = abl_data.get('full_model_gain_importance', {})
if gain_importance_full:
    gain_df_abl = pd.DataFrame(list(gain_importance_full.items()), columns=['Feature', 'Information Gain']).sort_values('Information Gain', ascending=False)

    def feat_color_type(f):
        if f in ['pagerank', 'degree_centrality']:
            return 'Graph Feature'
        elif f.startswith('spatial_'):
            return 'Spatial Proximity'
        return 'Base Feature'

    gain_df_abl['Type'] = gain_df_abl['Feature'].apply(feat_color_type)
    palette_feat = {'Graph Feature': '#2196F3', 'Spatial Proximity': '#4CAF50', 'Base Feature': '#9E9E9E'}

    plt.figure(figsize=(10, 6))
    sns.barplot(data=gain_df_abl, x='Information Gain', y='Feature', hue='Type', palette=palette_feat, dodge=False)
    plt.title('Feature Importance (Information Gain)\nFull Model: PageRank vs Spatial Proximity vs Base Features')
    plt.xlabel('Information Gain (Higher = More Important)')
    plt.legend(title='Feature Type', loc='lower right')
    plt.tight_layout()
    plt.show()

main_time_rows = []
for key, metrics in eval_data.items():
    if key in model_keys_to_skip:
        continue
    for fi, t in enumerate(metrics['training_time_seconds_folds']):
        main_time_rows.append({'Model': key, 'Fold': f'Fold {fi + 1}', 'Train Time (s)': t})
main_time_df = pd.DataFrame(main_time_rows)

print("\n" + "=" * 70)
print("PRINTOUT: Per-Fold Training Time — All 6 Models (Fixed Recording)")
print("=" * 70)
main_time_pivot = main_time_df.pivot(index='Fold', columns='Model', values='Train Time (s)')
print(main_time_pivot.to_string())
print("\nTotal (5-Fold Sum):")
print(main_time_pivot.sum().to_string())

plt.figure(figsize=(14, 6))
sns.barplot(data=main_time_df, x='Fold', y='Train Time (s)', hue='Model', palette='tab10')
plt.title('Per-Fold Training Time by Model (Actual Measured, seconds)')
plt.ylabel('Training Time (s)')
plt.legend(fontsize=8, loc='upper right')
plt.tight_layout()
plt.show()

main_total_time_rows = []
for key, metrics in eval_data.items():
    if key in model_keys_to_skip:
        continue
    main_total_time_rows.append({'Model': key, 'Total Train Time (s)': metrics['training_time_total']})
main_total_df = pd.DataFrame(main_total_time_rows)
plt.figure(figsize=(10, 6))
sns.barplot(data=main_total_df, x='Total Train Time (s)', y='Model', palette='tab10')
plt.title('Total Training Time per Model (5-Fold Sum, Fixed Recording)')
plt.xlabel('Total Training Time (s)')
plt.tight_layout()
plt.show()
```
